# 1. Arithmetic Operators

Used to perform basic mathematical operations.

| Operator | Description         | Example  | Output |
| -------- | ------------------- | -------- | ------ |
| `+`      | Addition            | `3 + 2`  | 5      |
| `-`      | Subtraction         | `5 - 1`  | 4      |
| `*`      | Multiplication      | `4 * 3`  | 12     |
| `/`      | Division            | `10 / 2` | 5.0    |
| `//`     | Floor Division      | `7 // 2` | 3      |
| `%`      | Modulus (remainder) | `7 % 3`  | 1      |
| `**`     | Exponentiation      | `2 ** 3` | 8      |


![Arithmetic Operators](https://miro.medium.com/v2/1*XRz6kN1VX6S1qRCr4GlUgw.png)


```python
# Examples of arithmetic operators
a = 10
b = 3

print(a + b)  # Addition
print(a - b)  # Subtraction
print(a * b)  # Multiplication
print(a / b)  # Division
print(a % b)  # Modulus
print(a ** b)  # Exponentiation
print(a // b)  # Floor Division
```

# 2. Comparison Operators
Used to compare two values. The result is True or False.


| Operator | Description      | Example  | Output |
| -------- | ---------------- | -------- | ------ |
| `==`     | Equal to         | `5 == 5` | True   |
| `!=`     | Not equal to     | `5 != 3` | True   |
| `>`      | Greater than     | `5 > 2`  | True   |
| `<`      | Less than        | `5 < 2`  | False  |
| `>=`     | Greater or equal | `5 >= 5` | True   |
| `<=`     | Less or equal    | `5 <= 6` | True   |

#  3. Logical Operators
Used to combine multiple conditions.


| Operator | Description                   | Example              | Output |
| -------- | ----------------------------- | -------------------- | ------ |
| `and`    | Returns True if both are true | `(x > 5 and x < 15)` | True   |
| `or`     | Returns True if one is true   | `(x < 5 or x > 15)`  | False  |
| `not`    | Reverses the result           | `not(x > 5)`         | False  |

![Logical Operators](https://miro.medium.com/v2/resize:fit:602/1*SDDa0gcAESWXdiiy3xs3PQ.png)

```python
x = 15

print(x > 5 and x < 10 )  # True and False = False
print(x < 5 or x > 10)    # False or True = True
print(not(x > 20))        # not(False) = True
```

## Scenario 1: Managing Student Grades

Imagine you're a teacher and you want to manage the grades of several students in your class. You have the grades for five students, and you need to perform several tasks:

1. Calculate the average grade.
2. Find the highest and lowest grade.
3. Add a new grade when a student submits their late assignment.
4. Remove the grade of a student who dropped the course.

To perform all of these tasks efficiently, we need a way to store multiple grades together in one place. Using separate variables for each grade would be inefficient and hard to manage. This is where Python's **list** data structure comes in handy. Lists allow us to store multiple pieces of data together in a single variable and easily manipulate them.

Let’s now explore how lists can help us solve this problem.


## Lists
- Lists are ordered collections of items.
- Lists can store items of any data type.
- Lists are mutable, meaning their contents can be changed.

![Lists](https://i.stechies.com/userfiles/images/python-list-index.webp)

```python
# Negative indexing in lists
fruits = ["apple", "banana", "cherry", "date"]
print(fruits[-1])  # Output: date
print(fruits[-2])  # Output: cherry
print(fruits[-3])  # Output: banana
print(fruits[-4])  # Output: apple

# Negative indexing in tuples
colors = ("red", "green", "blue", "yellow")
print(colors[-1])  # Output: yellow
print(colors[-2])  # Output: blue
print(colors[-3])  # Output: green
print(colors[-4])  # Output: red
```

```python
# Slicing lists using indexing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[1:4])  # Output: ['banana', 'cherry', 'date']
print(fruits[:3])  # Output: ['apple', 'banana', 'cherry']
print(fruits[2:])  # Output: ['cherry', 'date', 'elderberry']
print(fruits[-3:])  # Output: ['cherry', 'date', 'elderberry']

# Slicing tuples using indexing
colors = ("red", "green", "blue", "yellow", "purple")
print(colors[1:4])  # Output: ('green', 'blue', 'yellow')
print(colors[:3])  # Output: ('red', 'green', 'blue')
print(colors[2:])  # Output: ('blue', 'yellow', 'purple')
print(colors[-3:])  # Output: ('blue', 'yellow', 'purple')
```

## Scenario: Managing Student Grades
Let's solve the task!


1. Calculate the average grade.
2. Find the highest and lowest grade.
3. Add a new grade when a student submits their late assignment.
4. Remove the grade of a student who dropped the course (his grade was 90).

```python
# Creating a list of student grades
grades = [85, 90, 78, 92, 88]
average_grade = sum(grades) / len(grades)
print(f"The average grade is: {average_grade:.2f}")
hight_grades = max(grades)
low_grades = min(grades)
print(f"The highest grade is: {hight_grades} and the lowest grade is: {low_grades}")
late_grades = float(input("Enter the late grade: "))
grades.append(late_grades)
print(grades)
grades.remove(90)
print(f"final grades are: {grades}")
```
