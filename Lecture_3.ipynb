# Tuples 

- Tuples are ordered collections of items, similar to lists.
- Tuples are immutable, meaning their contents cannot be changed after creation.

## Scenario 2: Storing Patient Medical Information

In a healthcare system, you need to store important patient information such as their name, date of birth, and blood type. This data is crucial and should remain constant once recorded. To ensure that this data stays unmodified, we can use a **tuple**.

Tuples are useful because they are immutable, meaning the values they contain cannot be changed after they are created. This makes tuples perfect for storing sensitive information like patient details.


```python
jordan_flag_colors = ( "red" , "white" , "grden" , "black" )
#indexing
print(f"index of white: {jordan_flag_colors.index('white')}") 
print("first value of our : ",jordan_flag_colors[0])

print ("type of jordan_flag_colors:" ,type(jordan_flag_colors) )
#change any value in tuple by converting the tuple to list
print(f"tuple before change: {jordan_flag_colors}")
jordan_flag_colors = list(jordan_flag_colors)
print ("the new type of jordan_flag_colors:", type(jordan_flag_colors) )

jordan_flag_colors[2] = "green"

jordan_flag_colors = tuple(jordan_flag_colors)

print ( type(jordan_flag_colors) )

print("tupe after change:" ,jordan_flag_colors)
```

```python
jordan_flag_colors[0] = "pink"
```

```python
# Slicing lists using indexing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[1:4])  # Output: ['banana', 'cherry', 'date']
print(fruits[:3])  # Output: ['apple', 'banana', 'cherry']
print(fruits[2:])  # Output: ['cherry', 'date', 'elderberry']
print(fruits[-3:])  # Output: ['cherry', 'date', 'elderberry']

# Slicing tuples using indexing
colors = ("red", "green", "blue", "yellow", "purple")
print(colors[1:4])  # Output: ('green', 'blue', 'yellow')
print(colors[:3])  # Output: ('red', 'green', 'blue')
print(colors[2:])  # Output: ('blue', 'yellow', 'purple')
print(colors[-3:])  # Output: ('blue', 'yellow', 'purple')


#seq[start:stop:step]
print(fruits[1:4:2])  # Output: ['banana', 'cherry']
print(colors[1:4:2])  # Output: ('green', 'blue')
```

## Scenario  3 : Student Records

You are building a system to manage student data for a school. Each student has a unique ID, and you need to store details like their name, grade, and enrolled courses. These details will change over time, so you need a data structure that allows quick updates. A dictionary is ideal for this, as it lets you store and update information efficiently.

#### Task:
1. Use a dictionary where each student's ID is the key, and the value is another dictionary with their details (name, grade, and courses).
2. Access and display a student's grade by their ID.
3. Update a student's grade.
4. Add a new student with their details.


```python
# creating a dictionary
students_info = {
    "1" : {"name" : "ahmed" , "grade" : 90, "courses": ["python","java"] },
    "2" : {"name" : "mohamed" , "grade" : 80, "courses": ["python","c++"] },
    "3" : {"name" : "ali" , "grade" : 70, "courses": ["python","c"] }
}
# 1. Accessing student information by key
print(f"student grade: {students_info['1']['grade']}")
print(f"student grade: {students_info['2']['grade']}")
print(f"student grade: {students_info['3']['grade']}")
# 2. Modifying student information
students_info["1"]["grade"] = 100

print(f"student grade: {students_info['1']['grade']}")
print(f"student grade: {students_info['2']['grade']}")
print(f"student grade: {students_info['3']['grade']}")

# 3. Adding a new student
students_info["4"] = {"name" : "mohamed" , "grade" : 80, "courses": ["python","c++"] }

print(f"student grade: {students_info['1']}")
print(f"student grade: {students_info['2']}")
print(f"student grade: {students_info['3']}")
print(f"student grade: {students_info['4']}")
```

```python
# Creating and accessing dictionaries
person = {"name": "Alice", "age": 25, "city": "New York" }
print("our dictionary:" ,person)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
print("the keys of our dictionary:" ,person.keys())
print("the values of our dictionary:" ,person.values())
print("key-value pairs of our dictionary:" ,person.items())

# Accessing elements
print(person["name"])  # Output: Tasnim
print(person["age"])  # Output: 25
print(person["city"])  # Output: New York
```

## Sets
- Sets are unordered collections of unique items.
- Sets are mutable, but elements must be immutable.

![Sets](https://pynative.com/wp-content/uploads/2021/03/python-sets.jpg)

### Sets: Unique Course Enrollments

#### Scenario
You are managing a list of courses students have enrolled in. However, some courses appear more than once, as multiple students can enroll in the same course. To find out which courses are offered without duplicates, you need to use a set, which only stores unique items.

#### Task:
1. Convert the list of courses into a set to eliminate duplicates.
2. Add a new course to the set.


```python
courses = ["english", "arabic", "math", "science","math"]
courses_set= set(courses)
print(courses_set)
# Adding elements
courses_set.add("history")
print(courses_set)
```

```python
#  you cannot index a set
print(courses_set[0])
```

```python
courses_set.add([1,2])
```

```python
courses_set[0]=3
```

```python
# Creating and accessing sets
fruits = {"apple", "banana", "cherry"}
print(fruits)  # Output: {'apple', 'banana', 'cherry'}

# Adding elements
fruits.add("orange")
print(fruits)  # Output: {'apple', 'banana', 'cherry', 'orange'}

# Removing elements
fruits.remove("banana")
print(fruits)  # Output: {'apple', 'cherry', 'orange'}
```

 ## When to Use Each Data Structure

| Data Structure | Use Case |
|----------------|----------|
| **List**       | Use when you need an ordered collection of items that can be changed (mutable). Ideal for collections that may change size dynamically. |
| **Tuple**      | Use when you need an ordered collection of items that should not be changed (immutable). Suitable for fixed collections of related items. |
| **Dictionary** | Use when you need to associate unique keys with values. Great for fast lookups and dynamic data. |
| **Set**        | Use when you need an unordered collection of unique items. Perfect for membership tests and eliminating duplicates. |

| Feature              | **List**                  | **Tuple**              | **Dictionary**              | **Set**             |
| -------------------- | ------------------------- | ---------------------- | --------------------------- | ------------------- |
| Syntax               | `[1, 2, 3]`               | `(1, 2, 3)`            | `{"a": 1, "b": 2}`          | `{1, 2, 3}`         |
| Ordered              | ✅ Yes                     | ✅ Yes                  | ✅ Yes (Python 3.7+)         | ❌ No                |
| Mutable (Can change) | ✅ Yes                     | ❌ No                   | ✅ Yes                       | ✅ Yes               |
| Allows duplicates    | ✅ Yes                     | ✅ Yes                  | ❌ Keys: No<br>✅ Values: Yes | ❌ No                |
| Indexed              | ✅ Yes                     | ✅ Yes                  | ❌ (Keys used)               | ❌ No                |
| Key–Value pairs      | ❌ No                      | ❌ No                   | ✅ Yes                       | ❌ No                |
| Fast lookup          | ❌ Slower                  | ❌ Slower               | ✅ Very fast (by key)        | ✅ Very fast         |
| Use case             | Store a sequence of items | Fixed data (safe)      | Mapping related data        | Unique items only   |
| Example use          | Names, scores             | Coordinates, constants | User profiles               | Removing duplicates |

# Quiz

```python
def quiz():
    score = 0

    print("\n--- Data Structures Quiz ---\n")

    # Question 1: Lists
    print("Question 1: Which of the following statements about Python lists is TRUE?")
    print("a) Lists are immutable and cannot be changed after creation.")
    print("b) Lists are ordered collections and can contain duplicate items.")
    print("c) Lists store data in key-value pairs.")
    answer = input("Enter your answer (a/b/c): ")
    if answer.lower() == 'b':
        print("Correct! Lists are ordered, mutable, and allow duplicates.")
        score += 1
    else:
        print("Incorrect. The correct answer is b. Lists are ordered and allow duplicate items.")

    # Question 2: Tuples
    print("\nQuestion 2: What is the main characteristic that distinguishes a tuple from a list?")
    print("a) Tuples can only store numbers.")
    print("b) Tuples are unordered collections.")
    print("c) Tuples are immutable, meaning their elements cannot be changed after creation.")
    answer = input("Enter your answer (a/b/c): ")
    if answer.lower() == 'c':
        print("Correct! Tuples are immutable.")
        score += 1
    else:
        print("Incorrect. The correct answer is c. Tuples are immutable.")

    # Question 3: Dictionaries
    print("\nQuestion 3: How do you access values in a Python dictionary?")
    print("a) By their index (e.g., my_dict[0]).")
    print("b) By their unique key (e.g., my_dict['key_name']).")
    print("c) By iterating through the values directly.")
    answer = input("Enter your answer (a/b/c): ")
    if answer.lower() == 'b':
        print("Correct! Dictionaries use keys to access their corresponding values.")
        score += 1
    else:
        print("Incorrect. The correct answer is b. Values in a dictionary are accessed by their keys.")

    # Question 4: Sets
    print("\nQuestion 4: Which data structure automatically ensures that all its elements are unique?")
    print("a) List")
    print("b) Tuple")
    print("c) Set")
    answer = input("Enter your answer (a/b/c): ")
    if answer.lower() == 'c':
        print("Correct! Sets store only unique elements.")
        score += 1
    else:
        print("Incorrect. The correct answer is c. Sets are designed to store unique items.")

    print(f"\n--- Quiz End ---")
    print(f"Your final score is: {score}/4")
```

```python
quiz()
```
