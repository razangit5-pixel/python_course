#### Choosing a Valid Variable Name
- Must begin with a letter (a-z, A-Z) or an underscore (_).
- Cannot start with a number.
- Can only contain alpha-numeric characters and underscores (A-z, 0-9, and _).
- Case-sensitive (age, Age, and AGE are three different variables).
- Cannot be a reserved keyword in Python (e.g., `for`, `while`, `if`).


✅ Valid: name, _age, score1

❌ Invalid: 1name, my-name, print

##  input() Function

The input() function is used to get input from the user. It always returns the input as a string.

 Example:

```python
name =input("Enter your name")
print("Hello, " + name + "!")

age = int(input("How old are you?    ")) #'22'
print("Your age after 10 years" , age + 10)
print("Your age after 30 years" , age + 30)

print(name + ' you are nice')
```

If you want a number as input, you need to convert it using int() or float():

```python
age = int(input("Enter your age: "))
print("You will be", age + 1, "next year.")
```
